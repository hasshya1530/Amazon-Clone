5. Create a menu driven Python program to operate on a list of numbers which is provided by the user. The different operations to be provided are :

         a. Add all numbers

         b. Find the highest number

         c. Find the average of all the numbers

         d. Find the numbered with highest frequency in the list.

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))

while True:
    print("\nMENU")
    print("1. Add all numbers")
    print("2. Find highest number")
    print("3. Find average")
    print("4. Find number with highest frequency")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Sum =", sum(numbers))

    elif choice == 2:
        print("Highest number =", max(numbers))

    elif choice == 3:
        print("Average =", sum(numbers) / len(numbers))

    elif choice == 4:
        highest_frequency = 0
        most_frequent = numbers[0]

        for num in numbers:
            frequency = numbers.count(num)

            if frequency > highest_frequency:
                highest_frequency = frequency
                most_frequent = num

        print("Number with highest frequency =", most_frequent)

    elif choice == 5:
        print("Program ended")
        break

    else:
        print("Invalid choice")
```

6.  A school has many students studying in different standards. Each student has a roll number, name, marks in 4 subjects and standard. The school wants a system that can calculate his grade and promote him/her if his grade is greater than 'F'. The criteria for grade calculation is:

i. If percentage >=80,  grade  is  'A'
ii. if  60<=percentage<80, grade  is 'B'
iii. if 40<=percentage<60, grade is 'C'
iv. percentage<40, grade is 'F'

```python
class Student:
    def __init__(self, roll_no, name, marks, standard):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks
        self.standard = standard

    def calculate_grade(self):
        percentage = sum(self.marks) / 4

        if percentage >= 80:
            grade = "A"
        elif percentage >= 60:
            grade = "B"
        elif percentage >= 40:
            grade = "C"
        else:
            grade = "F"

        return grade

    def promote(self):
        grade = self.calculate_grade()

        if grade != "F":
            self.standard += 1
            print("Student promoted to standard", self.standard)
        else:
            print("Student is not promoted")


roll = int(input("Enter roll number: "))
name = input("Enter name: ")

marks = []
for i in range(4):
    marks.append(int(input("Enter marks: ")))

standard = int(input("Enter standard: "))

student = Student(roll, name, marks, standard)

print("Grade =", student.calculate_grade())
student.promote()
```

7. A bookshop has many books belonging to different categories like horror, romance, non-fiction etc. Each book has a name, ID,price, category and the author name. The bookshop wants an automated system with which it can get  the price of a particular book.
Also, the bookshop wants the system to update the price of a given category of books if needed by a given price. Implement the scenario using OOP in Python.

```python
class Book:
    def __init__(self, name, book_id, price, category, author):
        self.name = name
        self.book_id = book_id
        self.price = price
        self.category = category
        self.author = author


class BookShop:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def get_price(self, book_id):
        for book in self.books:
            if book.book_id == book_id:
                print("Price =", book.price)
                return

        print("Book not found")

    def update_category_price(self, category, new_price):
        for book in self.books:
            if book.category == category:
                book.price = new_price


shop = BookShop()

shop.add_book(Book("Dracula", 101, 500, "Horror", "Bram Stoker"))
shop.add_book(Book("It", 102, 600, "Horror", "Stephen King"))
shop.add_book(Book("The Notebook", 103, 400, "Romance", "Nicholas Sparks"))

shop.get_price(101)

shop.update_category_price("Horror", 700)

shop.get_price(101)
shop.get_price(102)
```
